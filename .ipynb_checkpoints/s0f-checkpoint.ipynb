```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('data/co2_gasmas/C2(0f)_2023_05_31_17-55.txt', sep=",", header=5)
data.columns = ["tau", "s0f"]
```

```python
plt.plot(data['tau'], data['s0f'])
```

```python
poly_coef = np.polyfit(data['tau'], data['s0f'], deg=3)
print(poly_coef)
```

```python
poly=np.poly1d(poly_coef)
print(poly)
```

```python
s2fit=poly(data['tau'])
```

```python
plt.plot(data['tau'], data['s0f'])
plt.plot(data['tau'], s2fit)
```

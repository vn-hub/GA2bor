```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from uncertainties import ufloat

# Definice kvadratické funkce
def func(x, a, b):
    return a * x**2 + b*x 

# Data
x_data = np.array([10, 25, 50, 100])
A_2f = np.array([0.370, 0.905, 1.574, 2.571])
H_2f = np.array([0.250, 0.606, 1.063, 1.793])
y_spread = np.array([0.011, 0.015, 0.042, 0.079])

# Fitování
best_fit_ab, covar = curve_fit(func, x_data, A_2f, sigma=y_spread, absolute_sigma=True)
sigma_ab = np.sqrt(np.diagonal(covar))

best_fit_ab2, covar2 = curve_fit(func, x_data, H_2f, sigma=y_spread, absolute_sigma=True)
sigma_ab2 = np.sqrt(np.diagonal(covar2))

# Nejistoty pomocí uncertainties
a = ufloat(best_fit_ab[0], sigma_ab[0])
b = ufloat(best_fit_ab[1], sigma_ab[1])
text_res = "Best-fit (linear) parameters: \na = {}\nb = {}".format(a, b)

a2 = ufloat(best_fit_ab2[0], sigma_ab2[0])
b2 = ufloat(best_fit_ab2[1], sigma_ab2[1])
text_res2 = "Best-fit (linear) parameters: \na = {}\nb = {}".format(a2, b2)

# Vytvoření grafu
fig, ax1 = plt.subplots()

color = 'tab:red'
ax1.set_xlabel('Carbon dioxide concentration [%vol.]', fontsize=14)
ax1.set_ylabel('2f-WMS peak amplitude [V]', color=color, fontsize=14)
ax1.tick_params(axis='y', labelcolor=color)

# Logaritmická osa x
ax1.set_xscale('log')

# Chybové úsečky a scatter plot pro A_2f
ax1.errorbar(x_data, A_2f, y_spread, fmt='none', linewidth=3, capsize=6, ecolor=color)
ax1.scatter(x_data, A_2f, marker='o', facecolor=color, alpha=0.9)

# Fitovací křivka pro A_2f
hires_x = np.logspace(np.log10(min(x_data)), np.log10(max(x_data)), 100)
ax1.plot(hires_x, func(hires_x, *best_fit_ab), color)

# Intervaly spolehlivosti
bound_upper = func(hires_x, *(best_fit_ab + sigma_ab))
bound_lower = func(hires_x, *(best_fit_ab - sigma_ab))
ax1.fill_between(hires_x, bound_lower, bound_upper, color=color, alpha=0.15)

ax2 = ax1.twinx()  # Druhá osa y
color2 = 'tab:blue'
ax2.set_ylabel('2f-WMS peak height [V]', color=color2, fontsize=14)
ax2.tick_params(axis='y', labelcolor=color2)

# Logaritmická osa x
ax2.set_xscale('log')

# Chybové úsečky a scatter plot pro H_2f
ax2.errorbar(x_data, H_2f, y_spread, fmt='none', linewidth=3, capsize=6, ecolor=color2)
ax2.scatter(x_data, H_2f, marker='o', facecolor=color2, alpha=0.9)

# Fitovací křivka pro H_2f
ax2.plot(hires_x, func(hires_x, *best_fit_ab2), color2)

# Intervaly spolehlivosti
bound_upper = func(hires_x, *(best_fit_ab2 + sigma_ab2))
bound_lower = func(hires_x, *(best_fit_ab2 - sigma_ab2))
ax2.fill_between(hires_x, bound_lower, bound_upper, color=color2, alpha=0.15)

# Nastavení rozsahu os
plt.xlim(8, 110)
plt.ylim(0, 5)

# Přidání textových popisků
plt.text(12, 2, text_res, color='tab:red', fontsize=12)
plt.text(20, 4, text_res2, color='tab:blue', fontsize=12)

plt.savefig("best-fit_unc_A_log.png", dpi=600, bbox_inches='tight')
plt.savefig("best-fit_unc_A_log.pdf", dpi=600, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Data
x_data = np.array([10, 25, 50, 100])
A_2f = np.array([0.370, 0.905, 1.574, 2.571])

# Definice funkce
def func(x, a, b):
    return a * x**2 + b*x 

# Fitování dat
best_fit_ab, _ = curve_fit(func, x_data, A_2f)

# Vytvoření grafu
fig, ax = plt.subplots()

ax.set_xscale("log")  # Logaritmická osa X
ax.set_yscale("log")  # Logaritmická osa Y

ax.scatter(x_data, A_2f, color="red", label="Data")  # Body
hires_x = np.logspace(np.log10(min(x_data)), np.log10(max(x_data)), 100)  # Logaritmické rozložení x
ax.plot(hires_x, func(hires_x, *best_fit_ab), color="blue", label="Fit")  # Fitovací křivka

plt.xlim(10, 100)
plt.ylim(0.1, 3)

# Popisky
ax.set_xlabel("CO2 concentration [%vol.]")
ax.set_ylabel("2f-WMS peak amplitude [V]")
ax.legend()

# Zobrazení
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Data
x_data = np.array([10, 25, 50, 100])
A_2f = np.array([0.370, 0.905, 1.574, 2.571])

# Definice funkce (kvadratická)
def func(x, a, b):
    return a * x**2 + b * x  

# Fitování dat
best_fit_ab, _ = curve_fit(func, x_data, A_2f)

# Vytvoření grafu
fig, ax = plt.subplots()

ax.set_xscale("log")  # Logaritmická osa X
ax.set_yscale("linear")  # Lineární osa Y (není nutné, je to výchozí)

ax.scatter(x_data, A_2f, color="red", label="Data")  # Body
hires_x = np.logspace(np.log10(min(x_data)), np.log10(max(x_data)), 100)  # Logaritmická interpolace x
ax.plot(hires_x, func(hires_x, *best_fit_ab), color="blue", label="Fit")  # Fitovací křivka

# Popisky
ax.set_xlabel("CO2 concentration [%vol.]")
ax.set_ylabel("2f-WMS peak amplitude [V]")
ax.legend()

# Zobrazení
plt.show()
```
